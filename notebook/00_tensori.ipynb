# PyTorch: i tensori

I tensori sono i "contenitori di numeri" su cui lavorano le reti neurali: come gli array NumPy, ma possono stare sulla GPU.

```python
import torch
import numpy as np
import pandas as pd
```

Un **tensore** è un array a più dimensioni. Un tensore a una dimensione è quindi un semplice array, cioè un vettore.
[https://pytorch.org/tutorials/](https://pytorch.org/tutorials)

```python
int_to_tensor = torch.tensor([10, 11, 12, 13])
print("Tensor object type after conversion: ", int_to_tensor.dtype)
print("Tensor object type after conversion: ", int_to_tensor.type())
```

Con lo stesso metodo `torch.tensor()` si converte una lista di decimali in un tensore di decimali.

```python
float_to_tensor = torch.tensor([10.0, 11.0, 12.0, 13.0])
print("Tensor object type after conversion: ", float_to_tensor.dtype)
print("Tensor object type after conversion: ", float_to_tensor.type())
```

Gli elementi della lista devono essere tutti dello stesso tipo. Si può anche scegliere il tipo del tensore: qui una lista di interi diventa un tensore di decimali.

```python
int_list_to_float_tensor = torch.FloatTensor([10, 11, 12, 13])
int_list_to_float_tensor.type()
print("Tensor  type after conversion: ", int_list_to_float_tensor.type())
```

I metodi `size()` e `ndimension()` danno la dimensione e il numero di dimensioni del tensore.

```python
print("Size of the int_list_to_float_tensor: ", int_list_to_float_tensor.size())
print("Dimensions of the int_list_to_float_tensor: ",int_list_to_float_tensor.ndimension())
```

Per cambiare la forma di un tensore si usa `view()`, con righe e colonne come argomenti.

```python
reshaped_tensor = int_list_to_float_tensor.view(4, 1)
print("Original size of the tensor: ", int_list_to_float_tensor.size())
print("New size of the tensor: ", reshaped_tensor.size())
```

`view()` ha cambiato la forma in `torch.Size([4, 1])`: 4 righe e 1 colonna.

Il numero di elementi deve restare lo stesso; con -1 (per esempio `view(-1, 1)`) PyTorch calcola da solo quella dimensione.

PyTorch converte anche gli array NumPy in tensori, con `torch.from_numpy`.

```python
numpy_arr = np.array([10.0, 11.0, 12.0, 13.0])
from_numpy_to_tensor = torch.from_numpy(numpy_arr)

print("dtype of the tensor: ", from_numpy_to_tensor.dtype)
print("type of the tensor: ", from_numpy_to_tensor.type())
```

E al contrario, un tensore torna array NumPy con `.numpy()`.

```python
tensor_to_numpy = from_numpy_to_tensor.numpy()
print("back to numpy from tensor: ", tensor_to_numpy)
print("dtype of converted numpy array: ", tensor_to_numpy.dtype)
```

# Da Pandas a tensore
Anche una serie Pandas si converte in tensore, passando dai suoi valori NumPy (`.values`).

```python
pandas_series=pd.Series([1, 0.2, 3, 13.1])
store_with_numpy=torch.from_numpy(pandas_series.values)
print("Stored tensor in numpy array: ", store_with_numpy)
print("dtype of stored tensor: ", store_with_numpy.dtype)
print("type of stored tensor: ", store_with_numpy.type())
```

`item()` restituisce un numero Python da un tensore, `tolist()` una lista.

```python
new_tensor=torch.tensor([10, 11, 12, 13])
print("the second item is",new_tensor[1].item())
tensor_to_list=new_tensor.tolist()
print('tensor:', new_tensor,"\nlist:",tensor_to_list)
```

Indici e sezioni funzionano come nelle liste Python: il primo indice è 0. Si usano le parentesi quadre.

```python
tensor_index = torch.tensor([0, 1, 2, 3])
print("Check value at index 0:",tensor_index[0])
print("Check value at index 3:",tensor_index[3])
```

Come per le liste, si possono prendere sezioni di un tensore e cambiarne i valori.

```python
example_tensor = torch.tensor([50, 11, 22, 33, 44])
sclicing_tensor = example_tensor[1:4]
print("example tensor : ", example_tensor)
print("subset of example tensor:", sclicing_tensor)
```

Cambiamo il valore in posizione 3:

```python
print("value at index 3 of example tensor:", example_tensor[3])
example_tensor[3] = 0
print("new tensor:", example_tensor)
```

# Minimo e massimo
`min()` e `max()` trovano il valore più piccolo e più grande.

```python
sample_tensor = torch.tensor([5, 4, 3, 2, 1])
min_value = sample_tensor.min()
max_value = sample_tensor.max()
print("check minimum value in the tensor: ", min_value)
print("check maximum value in the tensor: ", max_value)
```

# Media e deviazione standard
`mean()` e `std()` calcolano media e deviazione standard.

```python
mean_std_tensor = torch.tensor([-1.0, 2.0, 1, -2])
Mean = mean_std_tensor.mean()
print("mean of mean_std_tensor: ", Mean)
std_dev = mean_std_tensor.std()
print("standard deviation of mean_std_tensor: ", std_dev)
```

# Somma e prodotto
Somma e prodotto tra tensori si scrivono come tra numeri: elemento per elemento.

```python
a = torch.tensor([1, 1])
b = torch.tensor([2, 2])

add = a + b
multiply = a * b

print("addition of two tensors: ", add)
print("multiplication of two tensors: ", multiply)
```

Confrontiamo la velocità di **NumPy** e **PyTorch** nella moltiplicazione di matrici, con array e tensori 4D casuali.

Se non hai attivato la GPU: **Runtime → Cambia tipo di runtime → T4 GPU**. Senza GPU il confronto si fa sul processore.

```python
#!pip install torch torchvision torchaudio


# Check if CUDA is available
if torch.cuda.is_available():
    device = torch.device("cuda")
    print("CUDA is available. Using GPU.")
else:
    device = torch.device("cpu")
    print("CUDA is not available. Using CPU.")

# 4D arrays
array1 = np.random.rand(100, 100, 100, 100)
array2 = np.random.rand(100, 100, 100, 100)

# 4D tensors
tensor1 = torch.rand(100, 100, 100, 100).to(device)
tensor2 = torch.rand(100, 100, 100, 100).to(device)
```

```python
%%timeit
np.matmul(array1, array2)
```

```python
%%timeit
torch.matmul(tensor1, tensor2)
if device.type == "cuda": torch.cuda.synchronize()   # la GPU lavora in differita: aspettiamo che finisca
```
